```python
#1 write to check if a number entered by the user is odd or even?

num = int(input("Enter a Number : "))

if(num % 2 == 0):
    print("This number is even : ",num)
else:
    print("This number is odd : ",num)
```

```python
#2 write to FInd the greatest of 3 numbers entered by the user?

num1 = int(input("Enter 1st number : "))
num2 = int(input("Enter 2nd number : "))
num3 = int(input("Enter 3rd number : "))

if(num1 > num2 & num1 > num3):
    print(num1)
elif(num2 > num3):
    print(num2)
else:
    print(num3)
```

```python
#3 write to check if a number is a multiple of 7 or not?

num = int(input("Enter a Number : "))

if(num % 7 == 0):
    print("The number is multiple of 7 : ",num)
else:
    print("The number is not multiple of 7 : ",num)
```
