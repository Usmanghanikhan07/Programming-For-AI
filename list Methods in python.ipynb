```python
#List method in python


#Append  Function

list = [1,2,3,4,5,6]

list.append(10)

print(list)

```

```python
#Sort  Function

list = [5,6,3,6,7,8]

list.sort()

print(list)
```

```python
#Sort  Function

list = [1,2,3,4,5]

list.sort(reverse = True)

print(list)
```

```python
#Sort  Function in String or character through reverse function

list = ["C","A","H","E"]

list.sort()

print(list)
```

```python
#Reverse  Function

list = ["C","A","H","E"]

list.reverse()

print(list)
```

```python
#Remove  Function

list = [5,6,3,6,5,8]

list.remove(6)

print(list)
```
