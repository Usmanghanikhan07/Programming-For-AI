```python
# Arithmatic operators

a = 5
b = 2

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a % b)  # Reminder
print(a ** b)  #  a^b
```

```python
# Relational operators

a = 50
b = 20

print(a == b)
print(a != b)
print(a >= b)
print(a > b)
print(a <= b)
print(a < b)
```

```python
# Assignment operators 

num = 10
num %= 2

print ("num : ", num)
```

```python
# Logical operators 

val1 = True
val2 = False

print(val1 and val2)
print(val1 or val2)
print(not(val1 and val2))

```
